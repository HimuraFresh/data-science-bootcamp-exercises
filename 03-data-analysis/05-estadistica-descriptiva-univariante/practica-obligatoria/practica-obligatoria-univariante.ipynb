````python
# Primero dividir las variables por tipo, usando la función que programamos en los ejercicios

def card_tipo(df, umbral_categoria = 10, umbral_continua = 30):

    # Primera parte: Preparo el dataset con cardinalidades, % variación cardinalidad, y tipos
    df_temp = pd.DataFrame(
        [df.nunique(), df.nunique()/len(df) * 100, df.dtypes]
    )
    df_temp = df_temp.T   # Como nos da los valores de las columnas en columnas, y quiero que estas sean filas, la transpongo

    df_temp = df_temp.rename(columns = {0: "Card", 1: "%_Card", 2: "Tipo"})  # Cambio el nombre de la transposición anterior para que tengan más sentido

    # Corrección para cuando solo tengo un valor
    df_temp.loc[df_temp.Card == 1, "%_Card"] = 0.00

    # Creo la columna de sugerencia de tipo de variable, empiezo considerando todas categóricas pero podría haber empezado por cualquiera
    df_temp["tipo_sugerido"] = "Categorica"

    df_temp.loc[df_temp["Card"] == 2, "tipo_sugerido"] = "Binaria"
    df_temp.loc[df_temp["Card"] >= umbral_categoria, "tipo_sugerido"] = "Numerica discreta"
    df_temp.loc[df_temp["%_Card"] >= umbral_continua, "tipo_sugerido"] = "Numerica continua"

    return df_temp
````

````python
card_tipo(df_titanic, 10, 25)
````

````python
df_tip = card_tipo(df_titanic, 10, 25)

categoricas = []
numericas = []

for col, tipo in df_tip["tipo_sugerido"].items():
    if tipo in ["Categorica", "Binaria"]:
        categoricas.append(col)
    else:  # Numérica discreta o continua
        numericas.append(col)
````

````python
cols_a_quitar = ["survived", "pclass", "sibsp", "adult_male"]

for col in cols_a_quitar:
    if col in categoricas:
        categoricas.remove(col)
````

````python
# Obtener coeficientes de varianza (CVs):
def get_CV(df, columna):
    columna = [columna] if type(columna) == str else columna
    return df[columna].describe().T["std"]/df[columna].describe().T["mean"] * 100
````

````python
get_CV(df_titanic, "age")
````
